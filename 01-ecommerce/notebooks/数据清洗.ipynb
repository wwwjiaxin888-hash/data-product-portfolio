```python
orders_1=orders.drop_duplicates()
```

```python
print('去重后行数为：',len(orders_1))
```

```python
orders_1= orders.drop_duplicates().copy()   # 加上.copy()
# 再执行后面转换时间的代码就不会出警告
```

```python
orders_1['dt']=pd.to_datetime(orders_1['timestamp'],unit='s')+pd.Timedelta(hours=8)
#unit='s':告诉pandas， timestamp单位是秒 ，先把秒数转化为UTC标准时间
#+pd.Timedelta(hours=8)：UTC时间+8小时=北京时间 
#dt:存入新列：带年月日分秒的北京时间
```

```python
orders_1['date']=orders_1['dt'].dt.date
#.dt:pandas时间序列访问器，.date:只提取日期部分，其他舍弃
```

```python
orders_1[['timestamp','dt','date']].head()
#获取这几列的前五行，观察格式
```
