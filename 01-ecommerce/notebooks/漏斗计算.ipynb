```python
import pandas as pd
```

```python
orders_1=pd.read_csv(filepath_or_buffer="../output/data_clean.csv",encoding="utf-8") 
#用一个变量来接受读取的信息
```

```python
behavior_count=orders_1.groupby('behavior_type')["user_id"].nunique().reset_index(name="user_cnt")
#reset_index的作用：1.把原来的索引变成dataframe这普通的一列；2.给series 的‘值’那一列也变成dataframe的一列，
#name参数：专门给series的值命名（只有series才有）
```

```python
behavior_count
```

```python
pv_cnt=behavior_count[behavior_count['behavior_type']=='pv']["user_cnt"].iloc[0]
#没有加iloc[]的时候是series即带索引的一维序列，利用iloc[0]:取series第0号位置元素筛选
#也可选择.item():专门从只有一个元素的series里面提取原生python数值，更干净
```

```python
behavior_count['covert_rate']=round(behavior_count['user_cnt']/pv_cnt,2)
#pandas中series支持广播运算，即哪一整列/一个单独的数字，pandas会自动把列里面每个元素分别去除以这个固定数字，不需要写循环，不需要遍历每一行
#不可以写成behavior_count['user_cnt']/behavior_count['user_cnt'][0]这种行与行之间互相匹配的写法
```
