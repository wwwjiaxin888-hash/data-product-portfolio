```python
import pandas as pd
```

```python
df=pd.read_csv(filepath_or_buffer="../output/data_clean.csv",encoding="utf-8") 
#用一个变量来接受读取的信息
```

```python
# 只保留购买行为
df_buy = df[df["behavior_type"] == "buy"]
```

```python
# 按日期、品类，去重统计购买用户
agg_df = df_buy.groupby(["date","category_id"])["user_id"].nunique().reset_index()
```

```python
agg_df.columns = ["date","category_id","buy_user_cnt"]
```

```python
agg_df.to_excel("../output/buy_user_agg.xlsx", index=False)
```
